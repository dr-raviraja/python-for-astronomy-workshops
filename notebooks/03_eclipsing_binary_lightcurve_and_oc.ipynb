Students explore a periodic binary star light curve, identify primary and secondary eclipses, and examine how Observed-minus-Calculated ($O-C$) timing shifts behave when an orbital period changes.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(2026)

# Simulated contact/close binary light curve (Period = 0.345 d)
period = 0.345
t0 = 2460000.12
times = np.sort(rng.uniform(2460000.0, 2460004.0, 600))
phase = ((times - t0) % period) / period

# Smooth ellipsoidal + eclipse profile in relative magnitude
ellipsoidal = 0.04 * np.cos(4.0 * np.pi * phase)
primary_ecl = 0.38 * np.exp(-0.5 * ((phase - 0.0) / 0.055)**2)
secondary_ecl = 0.32 * np.exp(-0.5 * ((phase - 0.5) / 0.055)**2)

mag = 12.5 - ellipsoidal + primary_ecl + secondary_ecl + rng.normal(0.0, 0.012, len(times))

plt.figure(figsize=(8, 4))
plt.scatter(phase, mag, s=12, alpha=0.65, color='#1B365D', label='Simulated Photometry')
plt.gca().invert_yaxis()
plt.xlabel('Orbital Phase')
plt.ylabel('Apparent Magnitude')
plt.title('Eclipsing Binary Phase-Folded Light Curve')
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

```python
# Illustrate quadratic O-C curve (secular period change)
epochs = np.arange(-150, 151, 10)
q_term = 1.8e-9  # days / cycle^2
oc_days = q_term * (epochs**2) + rng.normal(0.0, 1.5e-5, len(epochs))

plt.figure(figsize=(7.5, 3.8))
plt.plot(epochs, oc_days * 86400.0, 'o-', markersize=4, color='#2B6CB0', label='O-C Residuals (seconds)')
plt.xlabel('Cycle Number (E)')
plt.ylabel('O - C (seconds)')
plt.title('Secular Period Variation in O-C Diagram')
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```
