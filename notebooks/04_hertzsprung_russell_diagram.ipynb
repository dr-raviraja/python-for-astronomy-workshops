Students load a sample catalog of well-known stars (`data/sample_hr_stars.csv`), inspect effective temperatures ($T_{\mathrm{eff}}$) and luminosities ($L / L_{\odot}$), and construct a Hertzsprung-Russell diagram.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/sample_hr_stars.csv')
print(f'Loaded {len(df)} sample stars across groups:', df['group'].unique().tolist())
df.head(10)
```

```python
group_styles = {
    'Main Sequence': {'color': '#2B6CB0', 'marker': 'o', 'size': 45},
    'Giant': {'color': '#D97706', 'marker': 's', 'size': 60},
    'Supergiant': {'color': '#DC2626', 'marker': '^', 'size': 85},
    'White Dwarf': {'color': '#6B7280', 'marker': 'D', 'size': 40},
}

plt.figure(figsize=(9, 6.2))

for grp, style in group_styles.items():
    sub = df[df['group'] == grp]
    plt.scatter(
        sub['teff_k'],
        sub['lum_lsun'],
        s=style['size'],
        marker=style['marker'],
        color=style['color'],
        alpha=0.85,
        label=f"{grp} (n={len(sub)})"
    )

for _, row in df.iterrows():
    plt.annotate(
        row['star_name'],
        (row['teff_k'], row['lum_lsun']),
        xytext=(5, 3),
        textcoords='offset points',
        fontsize=8
    )

plt.xscale('log')
plt.yscale('log')
plt.gca().invert_xaxis()
plt.xlim(32000, 2600)
plt.ylim(2e-4, 4e5)
plt.xlabel('Effective Temperature $T_{\mathrm{eff}}$ (K)')
plt.ylabel('Luminosity $L / L_{\odot}$')
plt.title('Sample Hertzsprung-Russell Diagram')
plt.legend(loc='lower left')
plt.grid(True, which='both', alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
# Verify radius scaling: R / R_sun = sqrt(L / L_sun) * (5778 / Teff)^2
df['calc_radius_rsun'] = np.sqrt(df['lum_lsun']) * ((5778.0 / df['teff_k']) ** 2)
df[['star_name', 'spectral_type', 'teff_k', 'lum_lsun', 'radius_rsun', 'calc_radius_rsun']].round(3)
```
