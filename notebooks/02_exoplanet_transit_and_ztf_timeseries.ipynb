In this lab, students analyse an exoplanet transit light curve (inspired by hot-Jupiter targets such as Kepler-8b), fold timestamps onto orbital phase, and estimate the planet-to-star radius ratio ($R_p / R_\star$) from transit depth.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(108)

# Simulated multi-night survey time series (15 days, P = 3.5225 days)
period = 3.5225
t0 = 2459000.40
time_bjd = np.sort(rng.uniform(2459000.0, 2459015.0, 900))

# Compute orbital phase centred at 0.0
phase = ((time_bjd - t0 + 0.5 * period) % period) / period - 0.5

# Box-shaped transit model: depth = 0.0095 (~0.95%), half-duration = 0.022 phase
true_depth = 0.0095
flux = np.ones_like(phase)
in_transit = np.abs(phase) <= 0.022
flux[in_transit] -= true_depth

# Add realistic photometric scatter (1.5 mmag)
flux_obs = flux + rng.normal(0.0, 0.0015, size=len(phase))

plt.figure(figsize=(8, 4))
plt.scatter(phase, flux_obs, s=10, alpha=0.55, color='#2B6CB0', label='Observed Relative Flux')
plt.axhline(1.0, color='#4A5568', linestyle='--', linewidth=1)
plt.xlim(-0.15, 0.15)
plt.xlabel('Orbital Phase')
plt.ylabel('Normalised Flux')
plt.title('Phase-Folded Exoplanet Transit Light Curve')
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

```python
# Estimate transit depth and planet-to-star radius ratio: depth = (Rp / R_star)^2
baseline_flux = np.median(flux_obs[np.abs(phase) > 0.04])
bottom_flux = np.median(flux_obs[np.abs(phase) < 0.015])

measured_depth = (baseline_flux - bottom_flux) / baseline_flux
radius_ratio = np.sqrt(max(measured_depth, 0.0))

# If host star radius is 1.48 Solar radii (1 Solar radius ~ 9.73 Jupiter radii)
r_star_rjup = 1.48 * 9.73
r_planet_rjup = radius_ratio * r_star_rjup

print(f'Measured Transit Depth: {measured_depth * 100:.3f}%')
print(f'Radius Ratio (Rp/R*):   {radius_ratio:.4f}')
print(f'Estimated Planet Radius: {r_planet_rjup:.2f} R_Jup')
```
