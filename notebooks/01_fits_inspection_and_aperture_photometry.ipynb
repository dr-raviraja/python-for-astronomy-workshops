In this notebook, students simulate a 2D CCD star field with sky background and Poisson noise, inspect radial brightness profiles, and measure stellar flux using circular aperture photometry.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
size = 120
y, x = np.mgrid[0:size, 0:size]

# Simulate sky background + 2D Gaussian target star at (60, 60)
sky_level = 120.0
x0, y0, fwhm, peak = 60.0, 60.0, 4.5, 2500.0
sigma = fwhm / 2.355
star = peak * np.exp(-((x - x0)**2 + (y - y0)**2) / (2 * sigma**2))
ccd_image = rng.poisson(sky_level + star).astype(float)

plt.figure(figsize=(5, 4.5))
plt.imshow(ccd_image, origin='lower', cmap='viridis')
plt.colorbar(label='Counts (ADU)')
plt.title('Simulated CCD Star Frame')
plt.xlabel('X Pixel')
plt.ylabel('Y Pixel')
plt.show()
```

```python
# Circular Aperture Photometry with Sky Annulus Subtraction
r = np.sqrt((x - x0)**2 + (y - y0)**2)
ap_mask = r <= 7.0
sky_mask = (r >= 12.0) & (r <= 18.0)

sky_per_pixel = np.median(ccd_image[sky_mask])
net_flux = np.sum(ccd_image[ap_mask] - sky_per_pixel)
inst_mag = -2.5 * np.log10(net_flux)

print(f'Estimated Sky Background: {sky_per_pixel:.2f} ADU/pixel')
print(f'Net Stellar Flux:         {net_flux:.1f} ADU')
print(f'Instrumental Magnitude:   {inst_mag:.3f} mag')
```
